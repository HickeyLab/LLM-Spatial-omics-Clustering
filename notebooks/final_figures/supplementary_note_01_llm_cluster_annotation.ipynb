# Supplementary Note 1 final: LLM-assisted cluster annotation

This notebook contains the source-rebuilt prompt, representative marker-summary input, and representative cell-type output diagram. The rendering code is transferred from make_supplementary_llm_annotation_figure.py; the requested headings are added above the three boxes in the notebook code.

The displayed input and output are representative schema examples. They are not presented as a recovered verbatim raw model response.


```python
from textwrap import fill

import matplotlib.pyplot as plt
from IPython.display import display
from matplotlib.patches import FancyArrowPatch, FancyBboxPatch

def draw_panel(axis, body, facecolor, fontsize, code=False):
    axis.set_axis_off()
    axis.add_patch(
        FancyBboxPatch(
            (0, 0),
            1,
            1,
            boxstyle='round,pad=0.018,rounding_size=0.025',
            facecolor=facecolor,
            edgecolor='#42617c',
            linewidth=1.1,
            transform=axis.transAxes,
            clip_on=False,
        )
    )
    axis.text(
        0.05,
        0.95,
        body,
        transform=axis.transAxes,
        ha='left',
        va='top',
        fontsize=fontsize,
        linespacing=1.27,
        fontfamily='DejaVu Sans Mono' if code else 'DejaVu Sans',
        color='#1c2730',
    )

plt.rcParams.update(
    {
        'font.family': 'Arial',
        'pdf.fonttype': 42,
        'ps.fonttype': 42,
    }
)

figure = plt.figure(figsize=(15, 8), facecolor='white')

prompt_paragraphs = [
    'You are an expert bioinformatician with deep knowledge of CODEX imaging and intestinal cell biology.',
    '1. Load the attached file ("cluster" : [marker1, marker2, ...])',
    '2. Assign each cluster one of the following cell types: Plasma Endothelial CD4+ T cell Smooth muscle Enterocyte Goblet M2 Macrophage Stroma CD8+ T Nerve CD66+ Enterocyte Lymphatic Cycling TA ICC Neutrophil B DC CD7+ Immune MUC1+ Enterocyte Neuroendocrine',
    '3. Return a dictionary in python in which every key is a cluster number (numbered from 0-n) and value is the predicted cell type for that given cluster.',
    'For instance 0 would be a key denoting cluster 0 and its value would be the predicted cell type.',
]
prompt_body = '\n\n'.join(fill(paragraph, width=49) for paragraph in prompt_paragraphs)

input_body = (
    '{\n'
    '  "0":["Cytokeratin","CD49f"],\n'
    '  "1":["CD66","Cytokeratin","CD49f"],\n'
    '  "2":["CD38","CD44","BCL2"],\n'
    '  "3":["CD7","CD3","Cytokeratin","CD8","CD49f"],\n'
    '  "4":["ITLN1","SOX9","Cytokeratin","CD49f"],\n'
    '  "5":["Cytokeratin"],\n'
    '  "6":["aSMA","CD49a"],\n'
    '  "7":["CD31","CD36","CollagenIV","CD34","HLA-DR"],\n'
    '  "8":["CD49a","CHGA","aDefensin5"],\n'
    '  "9":["Vimentin","MUC1","CHGA"],\n'
    '  ...\n'
    '}'
)

output_body = (
    '{\n'
    '  0: "Enterocyte",\n'
    '  1: "CD66+ Enterocyte",\n'
    '  2: "Plasma",\n'
    '  3: "CD8+ T",\n'
    '  4: "Goblet",\n'
    '  5: "Enterocyte",\n'
    '  6: "Smooth muscle",\n'
    '  7: "Endothelial",\n'
    '  8: "Neuroendocrine",\n'
    '  9: "Neuroendocrine",\n'
    '  ...\n'
    '}'
)

prompt_axis = figure.add_axes((0.045, 0.30, 0.27, 0.44))
input_axis = figure.add_axes((0.365, 0.30, 0.27, 0.44))
output_axis = figure.add_axes((0.685, 0.30, 0.27, 0.44))

draw_panel(prompt_axis, prompt_body, '#edf4f8', fontsize=8.0)
draw_panel(input_axis, input_body, '#f4f7f1', fontsize=8.0, code=True)
draw_panel(output_axis, output_body, '#f8f2ee', fontsize=8.0, code=True)

# These headings are the requested additions; they were not present in the original producer.
for axis, heading in (
    (prompt_axis, 'Prompt'),
    (input_axis, 'Example Input'),
    (output_axis, 'Example Output'),
):
    position = axis.get_position()
    figure.text(
        (position.x0 + position.x1) / 2,
        position.y1 + 0.035,
        heading,
        ha='center',
        va='bottom',
        fontsize=5,
        fontweight='normal',
        fontfamily='Arial',
        color='black',
    )

for start, end in [((0.318, 0.52), (0.358, 0.52)), ((0.638, 0.52), (0.678, 0.52))]:
    figure.add_artist(
        FancyArrowPatch(
            start,
            end,
            transform=figure.transFigure,
            arrowstyle='-|>',
            mutation_scale=18,
            linewidth=1.5,
            color='#42617c',
        )
    )

SUPPLEMENTARY_NOTE_1_FIGURE = figure
display(SUPPLEMENTARY_NOTE_1_FIGURE)
```
