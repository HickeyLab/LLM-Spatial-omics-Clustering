# Supplementary Note 1 — exact 47-marker request example

This entry point displays one frozen Leiden–Opus 5.5 means-plus-fractions request and its accepted final JSON response. It is a 75-cluster example from the prespecified illustrative pipeline, chosen by batch index before any outcome selection. All 47 markers and the original 28-label dictionary are included. Values use the scientific encoder's integer-thousandths representation. The request excludes reference labels and clustering-method identities. This notebook makes no endpoint calls and does not display hidden reasoning.

```python
from pathlib import Path
import json, hashlib
from IPython.display import display, Markdown, Code
ROOT=next(p for p in [Path.cwd(),*Path.cwd().parents] if (p/"configs/manuscript_inventory.json").exists())
DATA=ROOT/"data/frozen/annotation47_prompt_example_20261110"
manifest=json.loads((DATA/"manifest.json").read_text())
for name,record in manifest["files"].items():
    assert hashlib.sha256((DATA/name).read_bytes()).hexdigest()==record["sha256"]
metadata=json.loads((DATA/"metadata.json").read_text())
assert metadata["final_labels_match_accepted_bank"] and metadata["no_live_execution"]
display(Markdown("### Literal request body"))
display(Code((DATA/"request_body.json").read_text(),language="json"))
display(Markdown("### Accepted final assistant content"))
display(Code((DATA/"assistant_final.json").read_text(),language="json"))
display(Markdown("### Source hashes and selection"))
display(Code(json.dumps(metadata,indent=2),language="json"))
```
